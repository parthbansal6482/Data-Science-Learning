```python
import joblib
import pandas as pd

model = joblib.load(
    "severe_heatwave_model1.pkl"
)

threshold = 0.30
```

```python
farm_df = pd.read_csv("/Users/parthbansal/Satyukt Analytics/Heatwave/raster data/rajasthan_heatwave.csv")

farm_features = [
    'tmax_prev1',
    'tmax_prev2',
    'tmax_prev3',
    'rolling_3day_avg',
    'rolling_3day_max',                                                                                 
    'rolling_3day_std',
    'temp_change_3day',
    'departure'
]

farm_df = farm_df.dropna(
    subset=farm_features
)

farm_probabilities = model.predict_proba(
    farm_df[farm_features]
)[:, 1]

farm_predictions = (
    farm_probabilities > threshold
).astype(int)

farm_df['severe_heatwave_probability'] = (
    farm_probabilities * 100
)

farm_df['severe_heatwave_prediction'] = (
    farm_predictions
)


farm_df.to_csv("rajasthan_heatwave_predictions.csv",index=False)
```
