```python
import pandas as pd

# Load CSV
df = pd.read_csv("rajasthan_heatwave_predictions.csv")

# Remove column
df['f_sh_prob'] = (
    df.groupby(['latitude', 'longitude'])
    ['sh_prob']
    .shift(-1)
)

# Save updated CSV
df.to_csv("rajasthan_heatwave_predictions.csv", index=False)

print("Column removed successfully")
```
